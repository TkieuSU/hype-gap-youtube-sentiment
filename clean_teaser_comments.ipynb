```python
# Import library
import pandas as pd
```

# Initial Cleaning

```python
# Load dataset
teaser = pd.read_csv('./data/teaser_comments.csv')
debut = pd. read_csv('./data/debut_comments.csv')
```

### Data Exploration:

```python
teaser.info()
```

```python
debut.info()
```

```python
# Convert likeCount to int
teaser["likeCount"] = teaser["likeCount"].astype(int)
debut["likeCount"] = debut["likeCount"].astype(int)
```

After dropping relevant NaNs, our dataset has a total of 33,086 comments from teaser videos. I notice that there are 29,049 unique comments by 24,011 unqiue authors. This tells us that there are some authors that may post the same comment multiple times. To account for this, we will remove dpulicate comments that are written by the same author for the same video.

```python
# drop duplicates for rows that have the same video id, author, and text
teaser = teaser.drop_duplicates(subset=["video_id", "author", "text"], keep="first")
debut = debut.drop_duplicates(subset=["video_id", "author", "text"], keep="first")

# There is nas in debut comments- Specifically text variable
debut = debut.dropna()

print(teaser.info())
print(debut.info())
```

Now, we've removed 381 duplicates for comments that have the same content, written by the same author, and for the same video. That should put us a total of 32,705 teaser comments.

```python
len(teaser)
```

```python
# read the cleaned data into a new csv
teaser.to_csv('./data/teaser_comments.csv', index = False)
debut.to_csv('./data/debut_comments.csv', index = False)
```

# Basic Statistical Analysis

```python
# comment counts per group
group_counts = teaser["group"].value_counts()
print(group_counts)
```

I notice that groups that are more popular generally have higher comment counts

```python
# most liked comments
top_comments = teaser.sort_values("likeCount", ascending=False).head(10)
print(top_comments[["group", "likeCount", "text"]])
```

I notice that a majority of the comments with the most likes are positive, suggesting that it is likelt that the majority of teaser comments show positive sentiment. 

```python
# most common words
from collections import Counter
words = " ".join(teaser["text"]).lower().split()
Counter(words).most_common(20)
```

I notice while there are many stopwords here as the most common words, there are some words that show positive sentiment such as 'omg', 'love', and 'excited', suggesting that positive sentiment may be more prevalent in the teaser videos than negative sentiment.
